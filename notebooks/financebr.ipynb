```python
import datetime as dt
import yfinance as yf
import pandas as pd

inicio = dt.date(2018, 1, 1)
fim = dt.date(2024, 12, 31)

lista_acoes = ["BBDC4", "BBAS3", "VALE3", "PETR4", "CSAN3"]
tickers = [acao + ".SA" for acao in lista_acoes]

# Baixar dados de fechamento
dados = yf.download(tickers, start=inicio, end=fim)['Close']

# Ajustar colunas: remover o sufixo .SA
dados.columns = lista_acoes

# Transformar o índice Date em uma coluna normal
dados = dados.reset_index()

print(dados.head())
```

```python
dados.to_csv('b3_18_24.csv', index=False)
```
